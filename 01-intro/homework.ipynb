```python
asian_cars_filtered_column_df = df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]]
```

```python
asian_cars_filtered_column_df.head(7)
```

```python
X = asian_cars_filtered_column_df.head(7).to_numpy()
print(X)
```

```python
import numpy as np
XTX = np.matmul(X.T, X)
XTX.shape
print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv.shape
print(XTX_inv)
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
z= np.matmul(XTX_inv,X.T)
print(z)
```

```python
w = np.matmul(z,y)
print(w)
```
